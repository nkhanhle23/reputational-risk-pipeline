## Load Similarity Results and Apply Threshold

Load the similarity results from notebook 12 and filter by similarity threshold.

```python
# Load similarity validation results from notebook 12
SIMILARITY_RESULTS_PATH = 'results/phase i/12_keyword_validation_results_v4.csv'
keyword_df = pd.read_csv(SIMILARITY_RESULTS_PATH)

print(f"📊 Loaded {len(keyword_df)} candidate phrase-seed keyword matches")
print(f"   Categories: {keyword_df['category'].nunique()}")
print(f"   Unique candidate phrases: {keyword_df['generated_keyword'].nunique()}")
print(f"\n📈 Similarity score statistics:")
print(keyword_df['similarity_score'].describe())

# Select relevant columns for expanded dictionary
expanded_keyword_df = keyword_df[['category', 'generated_keyword']].copy()
expanded_keyword_df.rename(columns={'generated_keyword': 'keyword'}, inplace=True)

# Remove duplicates (same phrase may match multiple seed keywords)
expanded_keyword_df = expanded_keyword_df.drop_duplicates().reset_index(drop=True)

print(f"\n✅ Unique expanded keywords: {len(expanded_keyword_df)}")
print(f"\nBreakdown by category:")
print(expanded_keyword_df['category'].value_counts().sort_index())
```

```python
expanded_keyword_df['category'].value_counts()
```

```python
# Load original seed keywords from taxonomy (7 categories)
TAXONOMY_PATH = 'data/00_raw/taxonomy_sheet_v2.csv'
taxonomy_df = pd.read_csv(TAXONOMY_PATH)

# Parse seed keywords from taxonomy
original_keywords_list = []
for _, row in taxonomy_df.iterrows():
    category = row['category']
    keywords_str = row['keywords']
    # Split by newlines and clean
    keywords = [kw.strip() for kw in keywords_str.split('\n') if kw.strip()]
    for keyword in keywords:
        original_keywords_list.append({'category': category, 'keyword': keyword})

original_keywords_df = pd.DataFrame(original_keywords_list)

print(f"📊 Loaded {len(original_keywords_df)} seed keywords from taxonomy")
print(f"\nBreakdown by category:")
print(original_keywords_df['category'].value_counts().sort_index())

# Save the original keywords for reference
ORIGINAL_KEYWORDS_PATH = 'results/phase i/13_seed_keywords.csv'
original_keywords_df.to_csv(ORIGINAL_KEYWORDS_PATH, index=False)
print(f"\n💾 Saved seed keywords to: {ORIGINAL_KEYWORDS_PATH}")
```

```python
# Calculate total keywords
print(f"\n📊 Total Keywords:")
print(f"   Seed keywords: {len(original_keywords_df)}")
print(f"   Expanded keywords: {len(expanded_keyword_df)}")
print(f"   Combined: {len(original_keywords_df) + len(expanded_keyword_df)}")
```

```python
# Merge seed keywords with expanded keywords
merged_keywords_df = pd.concat([original_keywords_df, expanded_keyword_df], ignore_index=True)

# Remove duplicates (in case some expanded keywords match seed keywords exactly)
initial_count = len(merged_keywords_df)
merged_keywords_df = merged_keywords_df.drop_duplicates(subset=['category', 'keyword']).reset_index(drop=True)
duplicates_removed = initial_count - len(merged_keywords_df)

print(f"📋 Merging Results:")
```

```python
# Save the final dictionary
FINAL_DICTIONARY_PATH = 'results/phase i/13_final_dictionary_v4.csv'
merged_keywords_df.to_csv(FINAL_DICTIONARY_PATH, index=False)

print(f"\n💾 Final dictionary saved to: {FINAL_DICTIONARY_PATH}")
print(f"   Total keywords: {len(merged_keywords_df)}")
print(f"   Categories: {merged_keywords_df['category'].nunique()}")
print(f"\n🎯 Ready for classification in notebook 21!")
```
